## 1. What is pseudocode?

Pseudocode is an informal, human-readable description of an algorithm. It is not tied to the exact syntax of a programming language.

A good pseudocode solution should be:

- clear;
- ordered;
- finite;
- easy to translate into code;
- detailed enough that another person can understand the logic.


## 2. Sequence

### Problem
Given two numbers, calculate and display their sum.

### Pseudocode

```text
START
INPUT first_number
INPUT second_number
SET total = first_number + second_number
OUTPUT total
END
```


````python
def add_numbers(first_number, second_number):
    total = first_number + second_number
    return total

result = add_numbers(10, 5)
print("Sum:", result)

assert result == 15
````

## 3. Selection

### Problem
Determine whether a number is positive, negative, or zero.

### Pseudocode

```text
START
INPUT number

IF number > 0
    OUTPUT "Positive"
ELSE IF number < 0
    OUTPUT "Negative"
ELSE
    OUTPUT "Zero"
END IF

END
```


````python
def classify_number(number):
    if number > 0:
        return "Positive"
    elif number < 0:
        return "Negative"
    else:
        return "Zero"

tests = [8, -3, 0]
for value in tests:
    print(value, "->", classify_number(value))

assert classify_number(8) == "Positive"
assert classify_number(-3) == "Negative"
assert classify_number(0) == "Zero"
````

## 4. Selection with several conditions

### Problem
Convert a numeric score into a simple grade.

### Rules

- 90–100 → A
- 80–89 → B
- 70–79 → C
- 60–69 → D
- below 60 → F

### Pseudocode

```text
START
INPUT score

IF score >= 90
    SET grade = "A"
ELSE IF score >= 80
    SET grade = "B"
ELSE IF score >= 70
    SET grade = "C"
ELSE IF score >= 60
    SET grade = "D"
ELSE
    SET grade = "F"
END IF

OUTPUT grade
END
```


````python
def get_grade(score):
    if not 0 <= score <= 100:
        raise ValueError("Score must be between 0 and 100.")

    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

for score in [95, 84, 72, 63, 41]:
    print(score, "->", get_grade(score))

assert get_grade(95) == "A"
assert get_grade(84) == "B"
assert get_grade(41) == "F"
````

## 5. Iteration with a `for` loop

### Problem
Calculate the sum of all integers from 1 to `n`.

### Pseudocode

```text
START
INPUT n
SET total = 0

FOR number FROM 1 TO n
    SET total = total + number
END FOR

OUTPUT total
END
```


````python
def sum_to_n(n):
    total = 0

    for number in range(1, n + 1):
        total += number

    return total

print("Sum from 1 to 10:", sum_to_n(10))
assert sum_to_n(10) == 55
````

## 6. Iteration with a `while` loop

### Problem
Count down from a starting number to 1.

### Pseudocode

```text
START
INPUT number

WHILE number > 0
    OUTPUT number
    SET number = number - 1
END WHILE

OUTPUT "Done"
END
```


````python
def countdown(number):
    values = []

    while number > 0:
        values.append(number)
        number -= 1

    return values

print("Countdown:", countdown(5))
assert countdown(5) == [5, 4, 3, 2, 1]
````

## 7. Working with a list

### Problem
Find the largest number in a list without using `max()`.

### Pseudocode

```text
START
INPUT numbers

SET largest = first item in numbers

FOR each number IN numbers
    IF number > largest
        SET largest = number
    END IF
END FOR

OUTPUT largest
END
```


````python
def find_largest(numbers):
    if not numbers:
        raise ValueError("The list cannot be empty.")

    largest = numbers[0]

    for number in numbers[1:]:
        if number > largest:
            largest = number

    return largest

values = [4, 12, 3, 27, 9, 18]
print("Largest:", find_largest(values))

assert find_largest(values) == 27
````

## 8. Average of a list

### Problem
Calculate the mean of a list of numeric values.

### Pseudocode

```text
START
INPUT numbers
SET total = 0
SET count = 0

FOR each number IN numbers
    SET total = total + number
    SET count = count + 1
END FOR

SET average = total / count
OUTPUT average
END
```


````python
def calculate_average(numbers):
    if not numbers:
        raise ValueError("The list cannot be empty.")

    total = 0
    count = 0

    for number in numbers:
        total += number
        count += 1

    return total / count

grades = [70, 85, 90, 75, 80]
print("Average:", calculate_average(grades))

assert calculate_average(grades) == 80
````

## 9. Even and odd numbers

### Problem
Count how many even and odd numbers are present in a list.

### Pseudocode

```text
START
INPUT numbers
SET even_count = 0
SET odd_count = 0

FOR each number IN numbers
    IF number MOD 2 = 0
        SET even_count = even_count + 1
    ELSE
        SET odd_count = odd_count + 1
    END IF
END FOR

OUTPUT even_count
OUTPUT odd_count
END
```


````python
def count_even_odd(numbers):
    even_count = 0
    odd_count = 0

    for number in numbers:
        if number % 2 == 0:
            even_count += 1
        else:
            odd_count += 1

    return even_count, odd_count

numbers = [1, 2, 3, 4, 5, 6, 8]
even, odd = count_even_odd(numbers)

print("Even:", even)
print("Odd:", odd)

assert (even, odd) == (4, 3)
````

## 10. FizzBuzz

### Problem
For numbers from 1 to `n`:

- multiples of 3 → `Fizz`
- multiples of 5 → `Buzz`
- multiples of both → `FizzBuzz`
- otherwise → the number

### Pseudocode

```text
START
INPUT n

FOR number FROM 1 TO n
    IF number MOD 15 = 0
        OUTPUT "FizzBuzz"
    ELSE IF number MOD 3 = 0
        OUTPUT "Fizz"
    ELSE IF number MOD 5 = 0
        OUTPUT "Buzz"
    ELSE
        OUTPUT number
    END IF
END FOR

END
```


````python
def fizzbuzz(n):
    result = []

    for number in range(1, n + 1):
        if number % 15 == 0:
            result.append("FizzBuzz")
        elif number % 3 == 0:
            result.append("Fizz")
        elif number % 5 == 0:
            result.append("Buzz")
        else:
            result.append(number)

    return result

output = fizzbuzz(15)
print(output)

assert output[2] == "Fizz"
assert output[4] == "Buzz"
assert output[14] == "FizzBuzz"
````

## 11. Palindrome checker

### Problem
Determine whether a word reads the same forward and backward.

### Pseudocode

```text
START
INPUT text
SET cleaned_text = lowercase version of text
SET reversed_text = reversed cleaned_text

IF cleaned_text = reversed_text
    OUTPUT "Palindrome"
ELSE
    OUTPUT "Not a palindrome"
END IF

END
```


````python
def is_palindrome(text):
    cleaned_text = "".join(
        character.lower()
        for character in text
        if character.isalnum()
    )

    return cleaned_text == cleaned_text[::-1]

examples = ["level", "Python", "A man a plan a canal Panama"]

for example in examples:
    print(f"{example!r} -> {is_palindrome(example)}")

assert is_palindrome("level") is True
assert is_palindrome("Python") is False
assert is_palindrome("A man a plan a canal Panama") is True
````

## 12. Prime-number checker

### Problem
Determine whether an integer greater than 1 is prime.

### Pseudocode

```text
START
INPUT number

IF number <= 1
    OUTPUT false
    STOP
END IF

FOR divisor FROM 2 TO square root of number
    IF number MOD divisor = 0
        OUTPUT false
        STOP
    END IF
END FOR

OUTPUT true
END
```


````python
def is_prime(number):
    if number <= 1:
        return False

    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1

    return True

for number in range(1, 21):
    if is_prime(number):
        print(number, end=" ")

print()

assert is_prime(2) is True
assert is_prime(17) is True
assert is_prime(18) is False
````

## 13. Decomposition into functions

### Problem
Given a list of scores:

1. validate the scores;
2. calculate the average;
3. determine the overall result.

### Pseudocode

```text
START

FUNCTION validate_scores(scores)
    FOR each score IN scores
        IF score < 0 OR score > 100
            RETURN false
        END IF
    END FOR
    RETURN true
END FUNCTION

FUNCTION average(scores)
    SET total = sum of scores
    RETURN total / number of scores
END FUNCTION

FUNCTION final_result(scores)
    IF validate_scores(scores) = false
        RETURN "Invalid"
    END IF

    SET avg = average(scores)

    IF avg >= 60
        RETURN "Pass"
    ELSE
        RETURN "Fail"
    END IF
END FUNCTION

END
```


````python
def validate_scores(scores):
    return bool(scores) and all(0 <= score <= 100 for score in scores)

def average_score(scores):
    return sum(scores) / len(scores)

def final_result(scores):
    if not validate_scores(scores):
        return "Invalid"

    avg = average_score(scores)
    return "Pass" if avg >= 60 else "Fail"

student_scores = [72, 68, 80, 59, 75]

print("Scores:", student_scores)
print("Average:", round(average_score(student_scores), 2))
print("Result:", final_result(student_scores))

assert final_result(student_scores) == "Pass"
assert final_result([40, 50, 45]) == "Fail"
assert final_result([110, 60]) == "Invalid"
````

## 14. Final reflection

Pseudocode separates **problem solving** from programming syntax. The main process is:

1. identify inputs and expected outputs;
2. break the problem into small ordered steps;
3. decide where selection or repetition is needed;
4. write the algorithm in clear pseudocode;
5. translate each pseudocode step into Python;
6. test the implementation with representative cases.

Writing pseudocode first makes complex programming tasks easier to reason about, test, and communicate.
