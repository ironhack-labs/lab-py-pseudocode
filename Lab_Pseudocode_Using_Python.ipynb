## Overview

This exercise will help you practice writing pseudocode and translating it into Python code. Pseudocode is a way to describe algorithms in a human-readable format, which can then be implemented in programming languages.



### Exercise 1: Multiply Two Numbers

Write a pseudocode to read two numbers and multiply them together. Then, write the equivalent Python code.

### Pseudocode

```text
START
READ first_number
READ second_number
SET product = first_number * second_number
PRINT product
END
```

````python
def multiply_numbers(first_number, second_number):
    return first_number * second_number

result = multiply_numbers(6, 7)
print("Product:", result)

assert result == 42
````

### Exercise 2: Check Even or Odd

Write pseudocode that checks if a number is even or odd and prints the result. Then, write the equivalent Python code.

### Pseudocode

```text
START
READ number

IF number MOD 2 = 0
    PRINT "Even"
ELSE
    PRINT "Odd"
END IF

END
```

````python
def even_or_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

print("8 ->", even_or_odd(8))
print("5 ->", even_or_odd(5))

assert even_or_odd(8) == "Even"
assert even_or_odd(5) == "Odd"
````

### Exercise 3: Sum of First N Natural Numbers

Write pseudocode to calculate the sum of the first N natural numbers. Then, write the equivalent Python code.

### Pseudocode

```text
START
READ N
SET total = 0

FOR number FROM 1 TO N
    SET total = total + number
END FOR

PRINT total
END
```

````python
def sum_natural_numbers(n):
    total = 0

    for number in range(1, n + 1):
        total += number

    return total

result = sum_natural_numbers(10)
print("Sum of first 10 natural numbers:", result)

assert result == 55
````

### Exercise 4: Find Maximum of Three Numbers

Write pseudocode that finds the maximum of three numbers entered by the user. Then, write the equivalent Python code.

### Pseudocode

```text
START
READ number1
READ number2
READ number3

SET maximum = number1

IF number2 > maximum
    SET maximum = number2
END IF

IF number3 > maximum
    SET maximum = number3
END IF

PRINT maximum
END
```

````python
def maximum_of_three(number1, number2, number3):
    maximum = number1

    if number2 > maximum:
        maximum = number2

    if number3 > maximum:
        maximum = number3

    return maximum

result = maximum_of_three(12, 25, 18)
print("Maximum:", result)

assert result == 25
````

### Exercise 5: Count Vowels in a String

Write pseudocode to count the number of vowels in a given string and print the count. Then, write the equivalent Python code.

### Pseudocode

```text
START
READ text
SET count = 0

FOR each character IN text
    IF character is a vowel
        SET count = count + 1
    END IF
END FOR

PRINT count
END
```

````python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for character in text:
        if character in vowels:
            count += 1

    return count

text = "Ironhack"
result = count_vowels(text)

print(f"Number of vowels in {text!r}:", result)

assert result == 3
````

### Bonus Exercise 6: Reverse a String

Write pseudocode to reverse a given string and print it out.

### Pseudocode

```text
START
READ text
SET reversed_text = empty string

FOR each character IN text
    SET reversed_text = character + reversed_text
END FOR

PRINT reversed_text
END
```

````python
def reverse_string(text):
    reversed_text = ""

    for character in text:
        reversed_text = character + reversed_text

    return reversed_text

result = reverse_string("Python")
print("Reversed string:", result)

assert result == "nohtyP"
````

### Bonus Exercise 7: Print Multiplication Table of a Number

Write pseudocode to print the multiplication table of a given number up to 10.

### Pseudocode

```text
START
READ number

FOR multiplier FROM 1 TO 10
    SET result = number * multiplier
    PRINT number, "x", multiplier, "=", result
END FOR

END
```

````python
def multiplication_table(number):
    table = []

    for multiplier in range(1, 11):
        result = number * multiplier
        table.append((multiplier, result))

    return table

number = 5

for multiplier, result in multiplication_table(number):
    print(f"{number} x {multiplier} = {result}")

assert multiplication_table(5)[9] == (10, 50)
````

### Bonus Exercise 8: Count Words in a String

Write pseudocode that counts the words in a given string and prints the result.

### Pseudocode

```text
START
READ text
SPLIT text INTO words
SET count = number of items in words
PRINT count
END
```

### Bonus Exercise 9: Find Factorial of a Number

Write pseudocode that calculates the factorial of a given positive integer.

### Pseudocode

```text
START
READ number
SET factorial = 1

FOR value FROM 1 TO number
    SET factorial = factorial * value
END FOR

PRINT factorial
END
```

````python
def factorial(number):
    if number < 0:
        raise ValueError("Factorial is only defined for non-negative integers.")

    result = 1

    for value in range(1, number + 1):
        result *= value

    return result

print("5! =", factorial(5))

assert factorial(5) == 120
assert factorial(0) == 1
````

### Bonus Exercise 10: FizzBuzz Problem

Write pseudocode for the FizzBuzz problem where you print numbers from 1 to N but replace multiples of three with "Fizz" and multiples of five with "Buzz".

### Pseudocode

```text
START
READ N

FOR number FROM 1 TO N
    IF number MOD 15 = 0
        PRINT "FizzBuzz"
    ELSE IF number MOD 3 = 0
        PRINT "Fizz"
    ELSE IF number MOD 5 = 0
        PRINT "Buzz"
    ELSE
        PRINT number
    END IF
END FOR

END
```

````python
def fizzbuzz(n):
    output = []

    for number in range(1, n + 1):
        if number % 15 == 0:
            output.append("FizzBuzz")
        elif number % 3 == 0:
            output.append("Fizz")
        elif number % 5 == 0:
            output.append("Buzz")
        else:
            output.append(number)

    return output

result = fizzbuzz(15)
print(result)

assert result[2] == "Fizz"
assert result[4] == "Buzz"
assert result[14] == "FizzBuzz"
````
